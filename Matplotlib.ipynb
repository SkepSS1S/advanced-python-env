```python
#ex 12
import pandas as pd
import matplotlib.pyplot as plt


df = pd.DataFrame({
    "Name": ["Arystan", "Adik", "Arsen", "Mark"],
    "Score": [99, 92, 78, 90]
})

# Plot
plt.plot(df["Name"], df["Score"], marker='o')  
plt.xlabel("Students")
plt.ylabel("Scores")
plt.title("Student Scores Line Plot")
plt.show()
```

```python
# ex13
plt.bar(df["Name"], df["Score"])
plt.xlabel("Students")
plt.ylabel("Scores")
plt.title("Student Scores Bar Chart")
plt.show()
```
